```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
data=pd.read_csv("/ddc2.csv")
ws=data['ws']
wd=data['wd']
colors=plt.cm.viridis(2)
ax=plt.subplot(projection='polar')
ax.bar(ws,wd, width=0.5, bottom=0, color=colors, alpha=0.6)
ax.set_theta_zero_location('N')
ax.set_theta_direction(-1)
plt.savefig("windros.jpg", dpi=300)
plt.show()
```
